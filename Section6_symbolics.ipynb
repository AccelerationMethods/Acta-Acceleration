```python
import sympy as sp

def check(expr):
    return sp.simplify(expr)
```

## 1 Verification of Theorem 6.1

### 1.1 Statement and proof (as in the manuscript)

**Theorem 6.1. (SAGA)** Let $f_i\in\mathcal{F}_{m,L}(\mathbb{R}^d)$ for $i=1,\ldots,n$, with $0<m<L$, $f=\tfrac1n\sum_{i=1}^nf_i$, and $B_i(x)=f_i(x)-f_i(x_\star)-\langle\nabla f_i(x_\star),\,x-x_\star\rangle$. Let $k\geq0$, $x_k\in\mathbb{R}^d$ and stored points $\phi_{i,k}\in\mathbb{R}^d$ for $i=1,\ldots,n$, let $x_{k+1}$ and the $\phi_{i,k+1}$ be given by
$$
\left\{\begin{aligned}
&\text{sample } i_k \text{ uniformly in } \{1,\ldots,n\},\\
&d_k=\nabla f_{i_k}(x_k)-\nabla f_{i_k}(\phi_{i_k,k})+\tfrac1n\textstyle\sum_{j=1}^n\nabla f_j(\phi_{j,k}),\\
&x_{k+1}=x_k-\alpha\,d_k,\\
&\phi_{i_k,k+1}=x_k,\quad\phi_{j,k+1}=\phi_{j,k}\ \text{ for } j\neq i_k ,
\end{aligned}\right.
$$
with $\alpha=\tfrac1{3L}$, and let
$$
\Phi_k=\frac1n\sum_{i=1}^nB_i(\phi_{i,k})+c\,\|x_k-x_\star\|^2,\qquad c=\frac{1}{2\alpha(1-\alpha m)n}.
$$
It holds that:
$$
\mathbb{E}_k[\Phi_{k+1}]\leqslant\Big(1-\min\Big\{\frac1{4n},\frac m{3L}\Big\}\Big)\,\Phi_k ,
$$
where $\mathbb{E}_k$ is the expectation over $i_k$ conditionally on the past.

*Proof.* The proof consists of a weighted sum of the following inequalities:

- for $i=1,\ldots,n$, smoothness and strong convexity of $f_i$ between $x_\star$ and $x_k$, with weight $\lambda_1=\tfrac{2\alpha c(L-m)}{nL}$:

  $$
  f_i(x_\star)\geq f_i(x_k)+\langle\nabla f_i(x_k),\,x_\star-x_k\rangle+\tfrac m2\|x_k-x_\star\|^2+\tfrac{1}{2(L-m)}\|\nabla f_i(x_k)-\nabla f_i(x_\star)-m(x_k-x_\star)\|^2,
  $$

- for $i=1,\ldots,n$, smoothness of $f_i$ between $\phi_{i,k}$ and $x_\star$, with weight $\lambda_2=\tfrac{3Lc\alpha^2}{n}$:

  $$
  B_i(\phi_{i,k})\geq\tfrac1{2L}\|\nabla f_i(\phi_{i,k})-\nabla f_i(x_\star)\|^2,
  $$

- strong convexity of $f$ between $x_k$ and $x_\star$, with weight $\lambda_3=4mc\alpha^2$:

  $$
  f(x_k)-f_\star\leq\tfrac1{2m}\|\nabla f(x_k)\|^2,
  $$

- a bound on the second moment of $d_k$, with weight $\lambda_4=c\alpha^2$:

  $$
  \mathbb{E}_k\|d_k\|^2\leq3\,\mathbb{E}_k\|\nabla f_{i_k}(x_k)-\nabla f_{i_k}(x_\star)\|^2+\tfrac32\,\mathbb{E}_k\|\nabla f_{i_k}(\phi_{i_k,k})-\nabla f_{i_k}(x_\star)\|^2-2\|\nabla f(x_k)\|^2 .
  $$

  To obtain it, let $X=\nabla f_{i_k}(x_k)-\nabla f_{i_k}(x_\star)$ and $Y=\nabla f_{i_k}(\phi_{i_k,k})-\nabla f_{i_k}(x_\star)$, so that $\mathbb{E}_kX=\nabla f(x_k)$, as $\nabla f(x_\star)=0$, and $d_k-\nabla f(x_k)=(X-\mathbb{E}_kX)-(Y-\mathbb{E}_kY)$. Since $\mathbb{E}_kd_k=\nabla f(x_k)$, we have

  $$
  \mathbb{E}_k\|d_k\|^2=\|\nabla f(x_k)\|^2+\mathbb{E}_k\|d_k-\nabla f(x_k)\|^2,
  $$

  and Young's inequality $\|a-b\|^2\leq3\|a\|^2+\tfrac32\|b\|^2$ gives

  $$
  \begin{aligned}
  \mathbb{E}_k\|d_k\|^2&=\|\nabla f(x_k)\|^2+\mathbb{E}_k\|d_k-\nabla f(x_k)\|^2\\
  &\leq\|\nabla f(x_k)\|^2+3\,\mathbb{E}_k\|X-\mathbb{E}_kX\|^2+\tfrac32\,\mathbb{E}_k\|Y-\mathbb{E}_kY\|^2 .
  \end{aligned}
  $$

  The inequality follows from $\mathbb{E}_k\|X-\mathbb{E}_kX\|^2=\mathbb{E}_k\|X\|^2-\|\nabla f(x_k)\|^2$ and from $\mathbb{E}_k\|Y-\mathbb{E}_kY\|^2\leq\mathbb{E}_k\|Y\|^2$.

The weighted sum reads

$$
\begin{aligned}
 0\geq{}&\lambda_1\textstyle\sum_{i=1}^n\big[f_i(x_k)-f_i(x_\star)+\langle\nabla f_i(x_k),\,x_\star-x_k\rangle+\tfrac m2\|x_k-x_\star\|^2\\
 &\qquad\qquad+\tfrac{1}{2(L-m)}\|\nabla f_i(x_k)-\nabla f_i(x_\star)-m(x_k-x_\star)\|^2\big]\\
 &+\lambda_2\textstyle\sum_{i=1}^n\big[\tfrac1{2L}\|\nabla f_i(\phi_{i,k})-\nabla f_i(x_\star)\|^2-B_i(\phi_{i,k})\big]\\
 &+\lambda_3\big[f(x_k)-f_\star-\tfrac1{2m}\|\nabla f(x_k)\|^2\big]\\
 &+\lambda_4\big[\mathbb{E}_k\|d_k\|^2-3\,\mathbb{E}_k\|\nabla f_{i_k}(x_k)-\nabla f_{i_k}(x_\star)\|^2\\
 &\qquad\qquad-\tfrac32\,\mathbb{E}_k\|\nabla f_{i_k}(\phi_{i_k,k})-\nabla f_{i_k}(x_\star)\|^2+2\|\nabla f(x_k)\|^2\big],
\end{aligned}
$$

which, using the updates, the identity

$$
 \mathbb{E}_k\Big[\frac1n\sum_{i=1}^nB_i(\phi_{i,k+1})\Big]=\Big(1-\frac1n\Big)\frac1n\sum_{i=1}^nB_i(\phi_{i,k})+\frac1n\big(f(x_k)-f_\star\big),
$$

and $\alpha=\tfrac1{3L}$, can be reformulated exactly as

$$
 0\geq\mathbb{E}_k[\Phi_{k+1}]-\Phi_k+\frac1n\Big(1-\frac{1}{2(1-\alpha m)}\Big)\frac1n\sum_{i=1}^nB_i(\phi_{i,k})+\alpha m\,c\,\|x_k-x_\star\|^2 .
$$

As $\alpha m=\tfrac m{3L}\leq\tfrac13$ gives $1-\tfrac1{2(1-\alpha m)}\geq\tfrac14$, the two terms of the potential are contracted by different factors:

$$
 \mathbb{E}_k[\Phi_{k+1}]\leq\Big(1-\frac1{4n}\Big)\frac1n\sum_{i=1}^nB_i(\phi_{i,k})+\Big(1-\frac m{3L}\Big)\,c\,\|x_k-x_\star\|^2 .
$$

Since both terms of $\Phi_k$ are nonnegative, the right-hand side is at most $\max\{1-\tfrac1{4n},1-\tfrac m{3L}\}\,\Phi_k=\big(1-\min\{\tfrac1{4n},\tfrac m{3L}\}\big)\,\Phi_k$, which concludes the proof. $\square$

### 1.2 SymPy verification

The number $n$ of components is a parameter of the verification, set in the next cell ($n=3$ here; any $n\geq1$ can be used), and the symbols attached to the components are generated accordingly. The expectation over the sampled index is the average over the $n$ possible values of $i_k$. To verify the expression symbolically, one can define the following symbols, where the gradient of $f_n$ at $x_\star$ is determined by $\sum_{i=1}^n\nabla f_i(x_\star)=0$, and where the constant $c$ is kept as a symbol until the final check:

```python
L = sp.symbols('L')
m = sp.symbols('m')
alpha = sp.symbols('alpha')
c = sp.symbols('c')
n = 3                                                                   # number of components (any n >= 1)

xk = sp.symbols('x_k')
xs = sp.symbols('x_\\star')
phi = [sp.Symbol(f'phi_{{{i},k}}') for i in range(1, n + 1)]
gxk = [sp.Symbol(f'\\nabla{{}}f_{{{i}}}(x_k)') for i in range(1, n + 1)]
gphi = [sp.Symbol(f'\\nabla{{}}f_{{{i}}}(phi_{{{i},k}})') for i in range(1, n + 1)]
gs = [sp.Symbol(f'\\nabla{{}}f_{{{i}}}(x_\\star)') for i in range(1, n)]
gs = gs + [-sum(gs)]                                                    # the gradients at x_star sum to zero
fxk = [sp.Symbol(f'f_{{{i}}}(x_k)') for i in range(1, n + 1)]
fphi = [sp.Symbol(f'f_{{{i}}}(phi_{{{i},k}})') for i in range(1, n + 1)]
fs = [sp.Symbol(f'f_{{{i}}}(x_\\star)') for i in range(1, n + 1)]

E = lambda values: sum(values) / sp.Integer(len(values))                # expectation over a uniformly sampled index
B = lambda i, point, value: value - fs[i] - gs[i] * (point - xs)        # Bregman distance B_i(point), with value = f_i(point)

gradf_xk = E(gxk)                                                       # gradient of f at x_k
f_xk = E(fxk)                                                           # f(x_k)
f_star = E(fs)                                                          # f_star
d = [gxk[i] - gphi[i] + E(gphi) for i in range(n)]                      # d_k when i_k = i
xkp1 = [xk - alpha * d[i] for i in range(n)]                            # x_{k+1} when i_k = i
```

Now, one can write the four inequalities used in the proof (in the form "$\ldots\leqslant 0$"; the first two are lists, one inequality per component) and their weights to form the weighted sum:

```python
inequality1 = [fxk[i] - fs[i] + gxk[i] * (xs - xk) + m / 2 * (xk - xs) ** 2 + (gxk[i] - gs[i] - m * (xk - xs)) ** 2 / 2 / (L - m) for i in range(n)]
inequality2 = [(gphi[i] - gs[i]) ** 2 / 2 / L - B(i, phi[i], fphi[i]) for i in range(n)]
inequality3 = f_xk - f_star - gradf_xk ** 2 / 2 / m
inequality4 = (E([d[i] ** 2 for i in range(n)]) - 3 * E([(gxk[i] - gs[i]) ** 2 for i in range(n)])
               - sp.Rational(3, 2) * E([(gphi[i] - gs[i]) ** 2 for i in range(n)]) + 2 * gradf_xk ** 2)
lambda1 = 2 * alpha * c * (L - m) / n / L
lambda2 = 3 * L * c * alpha ** 2 / n
lambda3 = 4 * m * c * alpha ** 2
lambda4 = c * alpha ** 2

weighted_sum = lambda1 * sum(inequality1) + lambda2 * sum(inequality2) + lambda3 * inequality3 + lambda4 * inequality4
```

Similarly, one can define the potential value and its conditional expectation after the step (when $i_k=i$, the stored point $\phi_{i,k+1}$ becomes $x_k$ and the other stored points are unchanged), and form the target inequality

```python
potentialk = E([B(i, phi[i], fphi[i]) for i in range(n)]) + c * (xk - xs) ** 2
potentialkp1 = E([E([B(j, xk, fxk[j]) if j == i else B(j, phi[j], fphi[j]) for j in range(n)]) + c * (xkp1[i] - xs) ** 2 for i in range(n)])

target_inequality = (potentialkp1 - potentialk + (1 - 1 / (2 * (1 - alpha * m))) / n * E([B(i, phi[i], fphi[i]) for i in range(n)])
                     + alpha * m * c * (xk - xs) ** 2)
```

It remains to check that the expressions match for $\alpha=\tfrac1{3L}$ and $c=\tfrac{1}{2\alpha(1-\alpha m)n}$, which we check by verifying their difference is identically zero

```python
check((weighted_sum - target_inequality).subs(c, 1 / (2 * alpha * (1 - alpha * m) * n)).subs(alpha, 1 / (3 * L)))
```

As complements: for a general step size $\alpha$ and constant $c$, the same weighted sum can be reformulated exactly as $0\geq\mathbb{E}_k[\Phi_{k+1}]-\Phi_k+\big(\tfrac1n-3Lc\alpha^2\big)\tfrac1n\sum_{i=1}^nB_i(\phi_{i,k})+\alpha mc\|x_k-x_\star\|^2+\big(\tfrac{2\alpha c(L-m)}{L}+4mc\alpha^2-\tfrac1n\big)\big(f(x_k)-f_\star\big)+\big(\tfrac{\alpha c}{L}-3c\alpha^2\big)\mathbb{E}_k\|\nabla f_{i_k}(x_k)-\nabla f_{i_k}(x_\star)\|^2$, whose last two coefficients vanish for $\alpha=\tfrac1{3L}$ and $c=\tfrac1{2\alpha(1-\alpha m)n}$, for which $\tfrac1n-3Lc\alpha^2=\tfrac1n\big(1-\tfrac{1}{2(1-\alpha m)}\big)$; the identity $\mathbb{E}_k\big[\tfrac1n\sum_{i=1}^nB_i(\phi_{i,k+1})\big]=\big(1-\tfrac1n\big)\tfrac1n\sum_{i=1}^nB_i(\phi_{i,k})+\tfrac1n\big(f(x_k)-f_\star\big)$ used in the proof; the last step of the proof, where $1-\tfrac1{2(1-u)}-\tfrac14=\tfrac{1-3u}{4(1-u)}$ is nonnegative for $u=\alpha m\leq\tfrac13$; and the value of the constant of the potential for $\alpha=\tfrac1{3L}$, namely $c=\tfrac{9L}{2(3-m/L)n}$, as used in Corollary 6.2.

```python
table_term = E([B(i, phi[i], fphi[i]) for i in range(n)])
target_general = (potentialkp1 - potentialk + (sp.Rational(1, n) - 3 * L * c * alpha ** 2) * table_term + alpha * m * c * (xk - xs) ** 2
                  + (2 * alpha * c * (L - m) / L + 4 * m * c * alpha ** 2 - sp.Rational(1, n)) * (f_xk - f_star)
                  + (alpha * c / L - 3 * c * alpha ** 2) * E([(gxk[i] - gs[i]) ** 2 for i in range(n)]))
c_value = 1 / (2 * alpha * (1 - alpha * m) * n)

(check(weighted_sum - target_general),                                             # general alpha and c
 check(((sp.Rational(1, n) - 3 * L * c * alpha ** 2) - (1 - 1 / (2 * (1 - alpha * m))) / n).subs(c, c_value).subs(alpha, 1 / (3 * L))),
 check(E([E([B(j, xk, fxk[j]) if j == i else B(j, phi[j], fphi[j]) for j in range(n)]) for i in range(n)])
       - ((1 - sp.Rational(1, n)) * table_term + (f_xk - f_star) / n)),            # identity used in the proof
 sp.factor(1 - 1 / (2 * (1 - sp.Symbol('u'))) - sp.Rational(1, 4)),                # last step of the proof, with u = alpha m: nonnegative for u <= 1/3
 check(c_value.subs(alpha, 1 / (3 * L)) - 9 * L / (2 * (3 - m / L) * n)))          # constant of the potential for alpha = 1/(3L)
```
